## I.6 Eigenvalues and Eigenvectors

### Sample Problems

````julia
A = [.8 .3; .2 .7]
````

````julia
EL = eigvals(A)
````

````julia
EV = eigvecs(A)
````

Eigenvectors from Text

````julia
TV = [1 .6; -1 .4]
````

#### Are these the same?

If so, then there should exist a $c_i$  such that $EV_i * c_i = TV_i$, for $i = 1 \dots n$

Alternatively, find $C$ such thaat $EV \cdot C = TV$

Use Julia  `\` Solve operator 

````julia
C = EV\TV
round.(C,digits=3)
````

````julia
EV * C ≈ TV
````

## Square Matrix Classification

| Matrix Type | Type of | Definition | Example | Key Features |
|-------------|---------|------------|---------|--------------|
| **Square** | - | Equal number of rows and columns | $A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}$ | May have complex eigenvalues |
| **Normal** | Square | $A'A = AA'$ | $A = \begin{bmatrix} 1 & 1 \\ -1 & 1 \end{bmatrix}$ | Orthogonal eigenvectors, $A = Q\Lambda Q'$ |
| **Orthogonal** | Normal | $Q'Q = I$ | $Q = \begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}$ | $\|\lambda_i\| = 1$, preserves lengths, $Q^{-1} = Q'$ |
| **Identity** | Orthogonal | $I_{ii} = 1$, else 0 | $I = \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}$ | $\det(I) = 1$, all $\lambda_i = 1$ |
| **Permutation** | Orthogonal | One 1 per row and column, rest 0 | $P = \begin{bmatrix} 0 & 1 & 0 \\ 0 & 0 & 1 \\ 1 & 0 & 0 \end{bmatrix}$ | $\det(P) = \pm 1$, $\lambda$ are roots of unity |
| **Rotation** | Orthogonal | $Q'Q = I$, $\det(Q) = +1$ | $R = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}$ | Preserves orientation, $\|\lambda\| = 1$ |
| **Reflection** | Orthogonal | $Q'Q = I$, $\det(Q) = -1$ | $F = \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}$ | Reverses orientation, has $\lambda = -1$ |
| **Skew-symmetric** | Normal | $A = -A'$ | $A = \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix}$ | Imaginary or zero eigenvalues, diagonal is zero |
| **Symmetric** | Normal | $A = A'$ | $A = \begin{bmatrix} 2 & 1 \\ 1 & 3 \end{bmatrix}$ | Real eigenvalues, orthogonal eigenvectors |
| **Diagonal** | Symmetric | $a_{ij} = 0$ if $i \neq j$ | $D = \begin{bmatrix} 2 & 0 \\ 0 & 5 \end{bmatrix}$ | Eigenvalues on diagonal |
| **Positive Semidefinite** | Symmetric | $x'Ax \geq 0$ for all $x$ | $A = \begin{bmatrix} 1 & 1 \\ 1 & 1 \end{bmatrix}$ | $\lambda \geq 0$, pivots $\geq 0$ |
| **Positive Definite** | Positive Semidefinite | $x'Ax > 0$ for all $x \neq 0$ | $A = \begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix}$ | $\lambda > 0$, pivots $> 0$ |
| **Projection** | Symmetric | $P^2 = P = P'$ | $P = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 0 \end{bmatrix}$ | $\lambda = 0$ or $1$, rank $r$ equals trace |
| **Non-Normal** | Square | $A'A \neq AA'$ | $U = \begin{bmatrix} 2 & 3 \\ 0 & 5 \end{bmatrix}$ | Generally non-orthogonal eigenvectors |
| **Upper Triangular** | Non-Normal | $a_{ij} = 0$ if $i > j$ | $U = \begin{bmatrix} 2 & 3 \\ 0 & 5 \end{bmatrix}$ | Eigenvalues on diagonal |
| **Lower Triangular** | Non-Normal | $a_{ij} = 0$ if $i < j$ | $L = \begin{bmatrix} 2 & 0 \\ 3 & 5 \end{bmatrix}$ | Eigenvalues on diagonal |
| **Markov** | Non-Normal | $m_{ij} \geq 0$, each column sums to 1 | $M = \begin{bmatrix} 0.8 & 0.3 \\ 0.2 & 0.7 \end{bmatrix}$ | $\lambda_{\max} = 1$, steady state exists |
| **Bidiagonal** | Non-Normal | 2 adjacent nonzero diagonals | $B = \begin{bmatrix} 2 & 3 & 0 \\ 0 & 4 & 5 \\ 0 & 0 & 6 \end{bmatrix}$ | From SVD computation |
| **Tridiagonal** | Non-Normal | Main diagonal and 2 adjacent diagonals | $T = \begin{bmatrix} 2 & -1 & 0 \\ -1 & 2 & -1 \\ 0 & -1 & 2 \end{bmatrix}$ | Fast $O(n)$ algorithms, can be symmetric |
| **Toeplitz** | Non-Normal | Constant along diagonals | $T = \begin{bmatrix} 5 & 2 & 1 \\ 3 & 5 & 2 \\ 4 & 3 & 5 \end{bmatrix}$ | Convolution operator, $O(n^2)$ solve |
| **Circulant** | Toeplitz | Toeplitz with wraparound symmetry | $C = \begin{bmatrix} 5 & 3 & 2 \\ 2 & 5 & 3 \\ 3 & 2 & 5 \end{bmatrix}$ | Diagonalized by Fourier matrix, Normal |
