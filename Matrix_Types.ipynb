## Square Matrix Classification
| Matrix Type | Type of | Definition | Example | Key Features |
|-------------|---------|------------|---------|--------------|
| **Square** | - | Equal number of rows and columns | $A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}$ | May have complex eigenvalues |
| **Normal** | Square | $A'A = AA'$ | $A = \begin{bmatrix} 1 & 1 \\ -1 & 1 \end{bmatrix}$ | Orthogonal eigenvectors, $A = Q\Lambda Q'$. 2×2 forms: $\begin{bmatrix} a & b \\ b & d \end{bmatrix}$ or $\begin{bmatrix} a & b \\ -b & a \end{bmatrix}$ |
| **Orthogonal** | Normal | $Q'Q = I$ | $Q = \begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}$ | $\|\lambda_i\| = 1$, preserves lengths, $Q^{-1} = Q'$ |
| **Identity** | Orthogonal | $I_{ii} = 1$, else 0 | $I = \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}$ | $\det(I) = 1$, all $\lambda_i = 1$ |
| **Permutation** | Orthogonal | One 1 per row and column, rest 0 | $P = \begin{bmatrix} 0 & 1 & 0 \\ 0 & 0 & 1 \\ 1 & 0 & 0 \end{bmatrix}$ | $\det(P) = \pm 1$, $\lambda$ are roots of unity |
| **Rotation** | Orthogonal | $Q'Q = I$, $\det(Q) = +1$ | $R = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}$ | Preserves orientation, $\|\lambda\| = 1$ |
| **Reflection** | Orthogonal | $Q'Q = I$, $\det(Q) = -1$ | $F = \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}$ | Reverses orientation, has $\lambda = -1$ |
| **Skew-symmetric** | Normal | $A = -A'$ | $A = \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix}$ | Imaginary or zero eigenvalues, diagonal is zero |
| **Symmetric** | Normal | $A = A'$ | $A = \begin{bmatrix} 2 & 1 \\ 1 & 3 \end{bmatrix}$ | Real eigenvalues, orthogonal eigenvectors |
| **Diagonal** | Symmetric | $a_{ij} = 0$ if $i \neq j$ | $D = \begin{bmatrix} 2 & 0 \\ 0 & 5 \end{bmatrix}$ | Eigenvalues on diagonal |
| **Positive Semidefinite** | Symmetric | $x'Ax \geq 0$ for all $x$ | $A = \begin{bmatrix} 1 & 1 \\ 1 & 1 \end{bmatrix}$ | $\lambda \geq 0$, pivots $\geq 0$ |
| **Positive Definite** | Positive Semidefinite | $x'Ax > 0$ for all $x \neq 0$ | $A = \begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix}$ | $\lambda > 0$, pivots $> 0$ |
| **Projection** | Symmetric | $P^2 = P = P'$ | $P = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 0 \end{bmatrix}$ | $\lambda = 0$ or $1$, rank $r$ equals trace |
| **Non-Normal** | Square | $A'A \neq AA'$ | $U = \begin{bmatrix} 2 & 3 \\ 0 & 5 \end{bmatrix}$ | Generally non-orthogonal eigenvectors |
| **Upper Triangular** | Non-Normal | $a_{ij} = 0$ if $i > j$ | $U = \begin{bmatrix} 2 & 3 \\ 0 & 5 \end{bmatrix}$ | Eigenvalues on diagonal |
| **Lower Triangular** | Non-Normal | $a_{ij} = 0$ if $i < j$ | $L = \begin{bmatrix} 2 & 0 \\ 3 & 5 \end{bmatrix}$ | Eigenvalues on diagonal |
| **Markov** | Non-Normal | $m_{ij} \geq 0$, each column sums to 1 | $M = \begin{bmatrix} 0.8 & 0.3 \\ 0.2 & 0.7 \end{bmatrix}$ | $\lambda_{\max} = 1$, steady state exists |
| **Bidiagonal** | Non-Normal | 2 adjacent nonzero diagonals | $B = \begin{bmatrix} 2 & 3 & 0 \\ 0 & 4 & 5 \\ 0 & 0 & 6 \end{bmatrix}$ | From SVD computation |
| **Tridiagonal** | Non-Normal | Main diagonal and 2 adjacent diagonals | $T = \begin{bmatrix} 2 & -1 & 0 \\ -1 & 2 & -1 \\ 0 & -1 & 2 \end{bmatrix}$ | Fast $O(n)$ algorithms, can be symmetric |
| **Toeplitz** | Non-Normal | Constant along diagonals | $T = \begin{bmatrix} 5 & 2 & 1 \\ 3 & 5 & 2 \\ 4 & 3 & 5 \end{bmatrix}$ | Convolution operator, $O(n^2)$ solve |
| **Circulant** | Toeplitz | Toeplitz with wraparound symmetry | $C = \begin{bmatrix} 5 & 3 & 2 \\ 2 & 5 & 3 \\ 3 & 2 & 5 \end{bmatrix}$ | Diagonalized by Fourier matrix, Normal |

## Derivation: All 2×2 Normal Matrices

For $N = \begin{bmatrix} a & b \\ c & d \end{bmatrix}$ to be normal, we need $N^T N = N N^T$.

$$N^T N = \begin{bmatrix} a & c \\ b & d \end{bmatrix} \begin{bmatrix} a & b \\ c & d \end{bmatrix} = \begin{bmatrix} a^2 + c^2 & ab + cd \\ ab + cd & b^2 + d^2 \end{bmatrix}$$

$$N N^T = \begin{bmatrix} a & b \\ c & d \end{bmatrix} \begin{bmatrix} a & c \\ b & d \end{bmatrix} = \begin{bmatrix} a^2 + b^2 & ac + bd \\ ac + bd & c^2 + d^2 \end{bmatrix}$$

**Equating diagonal entries:**

$$a^2 + c^2 = a^2 + b^2 \quad \Rightarrow \quad c^2 = b^2$$

$$b^2 + d^2 = c^2 + d^2 \quad \text{(automatically satisfied)}$$

**Equating off-diagonal entries:**

$$ab + cd = ac + bd$$

**From $c^2 = b^2$:** Either $c = b$ or $c = -b$

---

#### Case 1: $c = b$

$$ab + bd = ab + bd \quad \checkmark$$

**Result:** $N = \begin{bmatrix} a & b \\ b & d \end{bmatrix}$ (symmetric)

---

#### Case 2: $c = -b$

$$ab - bd = -ab + bd$$

$$2ab = 2bd$$

$$a = d$$

**Result:** $N = \begin{bmatrix} a & b \\ -b & a \end{bmatrix}$ (rotation-scaling form)

---

#### Conclusion

All 2×2 normal matrices have the form:

$$N = \begin{bmatrix} a & b \\ b & d \end{bmatrix} \quad \text{or} \quad N = \begin{bmatrix} a & b \\ -b & a \end{bmatrix}$$

## Derivation: All 2×2 Real Orthogonal Matrices

For $Q = \begin{bmatrix} a & b \\ c & d \end{bmatrix}$ to be orthogonal, we need $Q^T Q = I$.

$$Q^T Q = \begin{bmatrix} a & c \\ b & d \end{bmatrix} \begin{bmatrix} a & b \\ c & d \end{bmatrix} = \begin{bmatrix} a^2 + c^2 & ab + cd \\ ab + cd & b^2 + d^2 \end{bmatrix} = \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}$$

**This gives three equations:**

$$a^2 + c^2 = 1 \quad \text{(1)}$$

$$b^2 + d^2 = 1 \quad \text{(2)}$$

$$ab + cd = 0 \quad \text{(3)}$$

---

#### From equation (1): Column 1 is a unit vector

$$a^2 + c^2 = 1 \quad \Rightarrow \quad (a, c) = (\cos\alpha, \sin\alpha) \text{ for some } \alpha$$

#### From equation (2): Column 2 is a unit vector

$$b^2 + d^2 = 1 \quad \Rightarrow \quad (b, d) = (\cos\beta, \sin\beta) \text{ for some } \beta$$

#### From equation (3): Columns are orthogonal

$$ab + cd = 0$$

$$\cos\alpha \cos\beta + \sin\alpha \sin\beta = 0$$

$$\cos(\alpha - \beta) = 0$$

$$\alpha - \beta = \pm\frac{\pi}{2}$$

---

#### Case 1: $\beta = \alpha - \frac{\pi}{2}$ (det = +1, rotation)

$$b = \cos\beta = \cos\left(\alpha - \frac{\pi}{2}\right) = \sin\alpha$$

$$d = \sin\beta = \sin\left(\alpha - \frac{\pi}{2}\right) = -\cos\alpha$$

**Result:**

$$Q = \begin{bmatrix} \cos\alpha & \sin\alpha \\ \sin\alpha & -\cos\alpha \end{bmatrix}$$

Let $\theta = -\alpha$ to get standard form:

$$Q = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix} \quad \text{(rotation by } \theta \text{)}$$

---

#### Case 2: $\beta = \alpha + \frac{\pi}{2}$ (det = -1, reflection)

$$b = \cos\beta = \cos\left(\alpha + \frac{\pi}{2}\right) = -\sin\alpha$$

$$d = \sin\beta = \sin\left(\alpha + \frac{\pi}{2}\right) = \cos\alpha$$

**Result:**

$$Q = \begin{bmatrix} \cos\alpha & -\sin\alpha \\ \sin\alpha & \cos\alpha \end{bmatrix}$$

This is a **reflection** (can also be written as rotation composed with reflection).

---

#### General Form of All 2×2 Orthogonal Matrices

**Rotations** ($\det(Q) = +1$):

$$Q = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}$$

**Reflections** ($\det(Q) = -1$):

$$Q = \begin{bmatrix} \cos\theta & \sin\theta \\ \sin\theta & -\cos\theta \end{bmatrix}$$

---

### Special Case: Integer Entries

For integer orthogonal matrices, we need:

$$a^2 + c^2 = 1 \quad \text{with } a, c \in \mathbb{Z}$$

$$b^2 + d^2 = 1 \quad \text{with } b, d \in \mathbb{Z}$$

**Integer solutions to $x^2 + y^2 = 1$:**

$$(x, y) \in \{(1, 0), (-1, 0), (0, 1), (0, -1)\}$$

This gives **8 integer orthogonal matrices:**

**Rotations** (4 matrices):

$$\begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}, \quad \begin{bmatrix} 0 & -1 \\ 1 & 0 \end{bmatrix}, \quad \begin{bmatrix} -1 & 0 \\ 0 & -1 \end{bmatrix}, \quad \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix}$$

(Identity, 90°, 180°, 270° rotations)

**Reflections** (4 matrices):

$$\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}, \quad \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}, \quad \begin{bmatrix} -1 & 0 \\ 0 & 1 \end{bmatrix}, \quad \begin{bmatrix} 0 & -1 \\ -1 & 0 \end{bmatrix}$$

These 8 matrices form the **dihedral group $D_4$** (symmetries of a square).
